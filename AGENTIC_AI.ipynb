LLM CREATION

```python
from langchain_google_genai import ChatGoogleGenerativeAI
llm=ChatGoogleGenerativeAI(
model="gemini-3.5-flash-lite",
temperature=0.5)
```

```python
question=input()
response=llm.invoke(question)
print(response.text)
```

**Prompt Template for LLM**

```python
from langchain_core.prompts import ChatPromptTemplate

prompts=ChatPromptTemplate([
    ('system','You are a helpful assistant.Explain things in simple and short language'),
    ('human',"{question}")
])
```

```python
chain=prompts|llm
llm_respone=chain.invoke({'question':question})
print(llm_respone.content)
```

TOOL **CREATION**

```python
from langchain_tavily import TavilySearch
tavily_search=TavilySearch(
    max_results= 4,
    topic="general"
)
```

AGENT **CREATION**

```python
from langchain.agents import create_agent
agent=create_agent(
    model=llm,
    tools=[tavily_search],
    system_prompt=( "You are a helpful assistant."
        "For questions about latest, current or recent information, "
        "search the web using tavily before answering"
        "Use the search results to prepare a clear and simple answer"
        "If the results are insufficent say no .. dont guess any answer"
)
   )
```

```python
agent_response=agent.invoke(
    {'messages':[{'role':'user','content':question}] }
    )
print(agent_response["messages"][-1].text)
```
